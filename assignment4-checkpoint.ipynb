```python
#addition by matrix
def addition():
    A=[[1,2],
       [3,4]]
    B=[[5,6],
       [7,8]]
    C=[[0,0],
       [0,0]]
    for i in range (len(A)):
        for j in range(len(A[0])):
            C[i][j]=A[i][j]+B[i][j]

    print("matrix A:")
    print(A)
    print("matrix B:")
    print(B)
    print("Addition of matrices :")
    print(C)

addition()
```
